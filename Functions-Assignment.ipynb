#Functions

1. What is the difference between a function and a method in Python?
   - Function: Defined independently using def (outside classes).
     - Called directly by name (e.g., add(2,3)).
     - Does not require self or cls as a first parameter.
   - Method: Defined inside a class and bound to an object/class.
     - Called using dot notation (e.g., obj.method()).
     - Usually takes self (instance) or cls (class) as the first parameter.

2. Explain the concept of function arguments and parameters in Python.
   - Parameters: Variables listed inside the function definition.
     - They act as placeholders to receive values when the function is called.
   - Arguments: Actual values passed to a function when it is called.
     - These values get assigned to the corresponding parameters.

3. What are the different ways to define and call a function in Python?
   - Regular Function: def add(a, b): return a + b
   - Lamda Function: add = lambda a, b: a + b

4. What is the purpose of the `return` statement in a Python function?
   - It sends a value back to the caller of the function.
   - It ends the function’s execution immediately.
   - Without return, a function gives None by default.

5. What are iterators in Python and how do they differ from iterables?
   - Iterators: An object that produces items one at a time using __next__().
     - Created from an iterable using iter().
   - Iterables: An object that can be looped over (like list, tuple, str).
     - Provides an __iter__() method that returns an iterator.

6. Explain the concept of generators in Python and how they are defined.
   - Generators are special iterators that yield values one at a time, using the yield keyword instead of return.
   - They don’t store the whole sequence in memory → more memory-efficient.
   - Can be paused and resumed, which makes them useful for handling large data streams.
   - def countdown(n):
    while n > 0:
        yield n
        n -= 1

7. What are the advantages of using generators over regular functions?
   - Memory Efficient: They yield values one by one, so they don’t store the whole sequence in memory.
   - Lazy Evaluation: Values are produced only when needed, making them faster for large datasets.
   - Clean & Simple Code: Easier to write iterators with yield instead of managing __iter__() and __next__().

8. What is a lambda function in Python and when is it typically used?
   - A small anonymous function defined using the lambda keyword (no def, no name).
   - Can take any number of arguments but must have a single expression.
   - Used for short, throwaway functions where defining a full function is unnecessary.
   - With functions like map(), filter(), and sorted(),When you need a quick inline function.

9. Explain the purpose and usage of the `map()` function in Python.
   - Applies a given function to each item in an iterable (like list, tuple).
   - Returns a map object (iterator), which can be converted to list/tuple.
   - Helps avoid explicit loops for element-wise operations.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
    - map(): Applies a function to each item of an iterable.
      - Returns a map object (can convert to list/tuple).
    - reduce(): Applies a function cumulatively to the items of an iterable, reducing it to a single value.
      - Returns the final reduced result.
    - filter(): Selects items from an iterable that satisfy a condition.
      - Returns a filter object.



```python
#1.  Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.
def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)
x = sum_even_numbers([1, 2, 3, 4, 5, 6])
print(x)
```

```python
#2. Create a Python function that accepts a string and returns the reverse of that string.
def reverse_string(string):
    return string[::-1]
x = reverse_string("hello")
print(x)
```

```python
#3. Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.
def square_numbers(numbers):
    return [num ** 2 for num in numbers]
x = square_numbers([1, 2, 3, 4, 5])
print(x)
```

```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

for num in range(1, 201):
    if is_prime(num):
        print(num, "is prime")

```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

for num in fibonacci(10):
    print(num, end=" ")

```

```python
#6.  Write a generator function in Python that yields the powers of 2 up to a given exponent.
def power_of_2(n):
    power = 1
    for _ in range(n):
        yield power
        power *= 2

for power in power_of_2(5):
    print(power)
```

```python
#7.  Implement a generator function that reads a file line by line and yields each line as a string.
def read_lines(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()

for line in read_lines('example.txt'):
    print(line)
```

```python
#8.  Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.
data = [(1, 5), (3, 2), (2, 8), (4, 1)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)
```

```python
#9.  Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.
def c_to_f(c):
    return (c * 9/5) + 32

celsius = [0, 20, 37, 100]

fahrenheit = list(map(c_to_f, celsius))

print(fahrenheit)
```

```python
#10.  Create a Python program that uses `filter()` to remove all the vowels from a given string.
def remove_vowels(string):
    vowels = 'aeiouAEIOU'
    return ''.join(filter(lambda x: x not in vowels, string))

string = "Hello, World!"
result = remove_vowels(string)
print(result)
```
